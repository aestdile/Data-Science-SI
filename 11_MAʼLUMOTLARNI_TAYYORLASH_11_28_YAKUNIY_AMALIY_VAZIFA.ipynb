#3) Har bir ustundagi boʻsh qiymatlarni aniqlang va ularni toʻldirish strategiyasini ishlab chiqing.

```python
"""
Asosiy topilmalar:
- price = -1 aslida bo’sh qiymat. Bu qatorlarning hammasida item ham bo’sh.
- Shu 876 ta qatordan 446 tasida quantity va order_total bor, ya’ni price = order_total / quantity formulasi bilan narxni aniq tiklash mumkin.
- Qolgan 430 ta qatorda price, quantity, order_total uchalasi ham yo’q.
- Har bir item faqat bitta kategoriya va bitta narxga ega, shuning uchun item bo’yicha category aniq tiklanadi.
- order_total = price * quantity formulasi barcha to’liq qatorlarda bajariladi.

---------------
| Strategiya: |
-------------------------------------------------------------------------------------------------------------------
Ustun                        |  Usul                                          |  Sabab
-------------------------------------------------------------------------------------------------------------------
price(-1)                    |  NaN ga o’girib, order_total / quantity;       |  formula aniq natija beradi (446 ta)
                             |  qolganini kategoriya medianasi bilan          |
--------------------------------------------------------------------------------------------------------------------
category                     | item bo’yicha lug’at                           |  item→category bir qiymatli (32 ta)
--------------------------------------------------------------------------------------------------------------------
item                         | (category, price) juftligi yagona taomga       |  taxmin qilmaymiz: masalan,
                             | mos kelsa shu taom; aks holda "Unknown"        |  Drinks + 3.0 uchta ichimlikka mos keladi (672 ta tiklanadi)
----------------------------------------------------------------------------------------------------------------------------------------------------
quantity                     | median (3)                                     |  1 dan 5 gacha tekis taqsimlangan, boshqa ma’lumot yo’q
------------------------------------------------------------------------------------------------------------------------------------------------
order_total                  | price * quantity                               |  formula bo'yicha
---------------------------------------------------------------------------------------------------------------------
payment_method               | "Unknown"                                      |  uchta usul deyarli teng uchraydi, eng ko’p uchraganini qo’yish noto’g’ri bo’lardi

"""

import pandas as pd
import numpy as np
import re

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"

df = pd.read_csv(url).drop(columns="Unnamed: 0")
df.columns = ["order_id", "customer_id", "category", "item", "price",
              "quantity", "order_total", "order_date", "payment_method"]

print(df.isna().sum()[lambda s: s > 0])
print("price = -1 soni:", (df["price"] == -1).sum())

df["item"] = df["item"].apply(
    lambda s: s if pd.isna(s) else re.sub(r"[^A-Za-z0-9() .]", "", s).strip().title())
df["price"] = df["price"].replace(-1, np.nan)

mask = df["price"].isna() & df["quantity"].notna() & df["order_total"].notna()
df.loc[mask, "price"] = df.loc[mask, "order_total"] / df.loc[mask, "quantity"]

item_cat = df.dropna(subset=["item", "category"]).groupby("item")["category"].agg(lambda s: s.mode()[0])
mask = df["category"].isna() & df["item"].notna()
df.loc[mask, "category"] = df.loc[mask, "item"].map(item_cat)

known = df.dropna(subset=["item", "category", "price"])
pair = known.groupby(["category", "price"])["item"].agg(["nunique", lambda s: s.mode()[0]])
pair.columns = ["n", "item"]
unique_pair = pair[pair["n"] == 1]["item"]
mask = df["item"].isna() & df["category"].notna() & df["price"].notna()
keys = pd.MultiIndex.from_frame(df.loc[mask, ["category", "price"]])
df.loc[mask, "item"] = unique_pair.reindex(keys).values
df["item"] = df["item"].fillna("Unknown")

cat_median = df.groupby("category")["price"].median()
mask = df["price"].isna()
df.loc[mask, "price"] = df.loc[mask, "category"].map(cat_median)
df["price"] = df["price"].fillna(df["price"].median())

mask = df["quantity"].isna()
df["quantity"] = df["quantity"].fillna(df["quantity"].median())
df.loc[mask, "order_total"] = df.loc[mask, "price"] * df.loc[mask, "quantity"]

df["category"] = df["category"].fillna("Unknown")
df["payment_method"] = df["payment_method"].fillna("Unknown")

print("Yakuniy NaN soni:", df.isna().sum().sum())
print("item Unknown:", (df["item"] == "Unknown").sum())
print("order_total == price*quantity:", np.isclose(df.order_total, df.price * df.quantity).mean())
```

#9) Manfiy narx yoki 0 ga teng narx qiymatlarini toping va ularni qanday oʻzgartirish kerakligini hal qiling.

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"
df = pd.read_csv(url).drop(columns="Unnamed: 0")
df.columns = ["order_id", "customer_id", "category", "item", "price",
              "quantity", "order_total", "order_date", "payment_method"]

print("Manfiy narxlar:", (df["price"] < 0).sum())
print("Nolga teng narxlar:", (df["price"] == 0).sum())
print("Manfiy qiymatlar:", df.loc[df["price"] < 0, "price"].unique())

bad = df["price"] <= 0
print("quantity va order_total bor:", (bad & df["quantity"].notna() & df["order_total"].notna()).sum())
print("Uchalasi ham yo'q:", (bad & df["quantity"].isna() & df["order_total"].isna()).sum())

df["price_fixed"] = False
df.loc[bad, "price"] = np.nan

can = df["price"].isna() & df["quantity"].notna() & df["order_total"].notna()
df.loc[can, "price"] = df.loc[can, "order_total"] / df.loc[can, "quantity"]
df.loc[can, "price_fixed"] = True

cat_median = df.groupby("category")["price"].median()
rest = df["price"].isna()
df.loc[rest, "price"] = df.loc[rest, "category"].map(cat_median)
df.loc[rest, "price_fixed"] = True
df["price"] = df["price"].fillna(df["price"].median())

print("Narxi <= 0:", (df["price"] <= 0).sum())
print("Narxi NaN:", df["price"].isna().sum())
print("Tuzatilgan narxlar:", df["price_fixed"].sum())
print(sorted(df["price"].unique()))
```

#10) Boʻsh toifalar (kategoriya) mavjudligini tekshiring va toʻgʻri kategoriya berish strategiyasini ishlab chiqing.

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"
df = pd.read_csv(url).drop(columns="Unnamed: 0")
df.columns = ["order_id", "customer_id", "category", "item", "price",
              "quantity", "order_total", "order_date", "payment_method"]

print("category NaN:", df["category"].isna().sum())
print("Bo'sh matn:", (df["category"].astype(str).str.strip() == "").sum())
print(df["category"].value_counts(dropna=False))

miss = df["category"].isna()
print("item ma'lum:", (miss & df["item"].notna()).sum())
print("item ham yo'q:", (miss & df["item"].isna()).sum())

item_clean = (df["item"].str.replace(r"[?#]", "", regex=True)
              .str.replace(r"\s*\([\d.]+\)", "", regex=True)
              .str.strip().str.title())
item_cat = (df.assign(item_clean=item_clean)
              .dropna(subset=["item_clean", "category"])
              .groupby("item_clean")["category"].agg(lambda s: s.mode()[0]))

mask = df["category"].isna() & item_clean.notna()
df.loc[mask, "category"] = item_clean[mask].map(item_cat)
print("1) item orqali tiklandi:", mask.sum(), "| qoldi:", df["category"].isna().sum())

eff_price = df["price"].where(df["price"] > 0, df["order_total"] / df["quantity"])
known = df.assign(eff_price=eff_price).dropna(subset=["category", "eff_price"])
pc = known.groupby("eff_price")["category"].agg(["nunique", lambda s: s.mode()[0]])
pc.columns = ["n", "category"]
unique_price = pc[pc["n"] == 1]["category"]

mask = df["category"].isna() & eff_price.notna()
df.loc[mask, "category"] = eff_price[mask].map(unique_price)
print("2) narx orqali qoldi:", df["category"].isna().sum())

df["category"] = df["category"].fillna("Unknown")
print(df["category"].value_counts())
print("NaN soni:", df["category"].isna().sum())
```

#18) Mahsulotlar kategoriyalari bo‘yicha tahlil qiling va eng ko‘p sotiladigan kategoriyalarni aniqlang.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/alishermutalov/praktikum-datasets/refs/heads/praktikum/restaurant_sales.csv"
df = pd.read_csv(url).drop(columns="Unnamed: 0")
df.columns = ["order_id", "customer_id", "category", "item", "price",
              "quantity", "order_total", "order_date", "payment_method"]

report = df.groupby("category").agg(
    orders=("order_id", "count"),
    quantity=("quantity", "sum"),
    revenue=("order_total", "sum"),
    avg_order=("order_total", "mean")).round(2)
report["share_%"] = (report["revenue"] / report["revenue"].sum() * 100).round(1)

print(report.sort_values("quantity", ascending=False))
```

#19) Har bir mijozning oʻrtacha xarid miqdorini (`groupby("customer_id")`) hisoblang.
