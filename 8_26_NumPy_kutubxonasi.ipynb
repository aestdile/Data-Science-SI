#### 28. Ixtiyoriy kichik massivlar yarating, ular yigʻindisini `np.sum` funksiyasidanda tezroq hisoblovchi dastur tuzing.


`💡Ishora` np.add.reduce

````python
import numpy as np
import time

a = np.array([1, 2, 3, 4, 5])

start = time.perf_counter()
sum1 = np.sum(a)
time1 = time.perf_counter() - start

start = time.perf_counter()
sum2 = np.add.reduce(a)
time2 = time.perf_counter() - start

print("np.sum() natijasi:", sum1)
print("np.add.reduce() natijasi:", sum2)
print("np.sum() vaqti:", time1)
print("np.add.reduce() vaqti:", time2)
````

#### 31. Ixtiyoriy sonlardan tashkil topgan massiv yarating, unga oʻxshash qiymatlar kiriting. Soʻng, oʻxshash qiymatlarni topuvchi dastur tuzing.


`💡Ishora` np.bincount, argmax

````python
import numpy as np
a = np.array([1, 2, 3, 2, 4, 2, 5, 3, 3, 3])
print("Massiv:", a)
print("\nSonlarning takrorlanish soni:")
counts = np.bincount(a)
for i in range(len(counts)):
    print(f"{i} soni {counts[i]} marta takrorlangan")
most_common = np.argmax(counts)

print(f"\nEng ko'p takrorlangan son: {most_common}")
print(f"Takrorlanishlar soni: {counts[most_common]}")
````

#### BONUS topshiriq

`💡Oʻzingiz javob berishni istagan savolni kiriting va unga javob bering.`

````python
"""
-----      Aptnake Corporation — Cyber Attack Analytics -----

NumPy yordamida 2024–2025-yillardagi kiberhujumlarni tahlil qilish

Aptnake Corporation 2024 va 2025-yillarda sodir bo‘lgan kiberhujumlar soni hamda har bir hujumdan kelib chiqqan moliyaviy zararni tahlil qilmoqchi. Hujumlar va zarar summalari alohida NumPy datasetlarida saqlanadi.
Masala shartlari:
1. 2024 va 2025-yillar uchun alohida hujumlar datasetlarini yarating.
2. Har bir yil uchun hujumlar sonini aniqlang.
3. Har bir yil bo‘yicha jami moliyaviy zararni hisoblang.
4. Bitta hujumga to‘g‘ri keladigan o‘rtacha zararni toping.
5. Eng katta va eng kichik zararlarni aniqlang.
6. Qaysi yilda jami zarar ko‘proq bo‘lganini aniqlang.
7. 2024-yildan 2025-yilga jami zarar necha foizga o‘zgarganini hisoblang.
8. Zarar summasi $10,000 dan yuqori bo‘lgan hujumlarni ajrating va sanang.
9. Eng katta zarar keltirgan hujumning tartib raqamini aniqlang.
Valyuta: AQSh dollari (USD).
"""

#----------------------------------------------------------------#
print("-----      Aptnake Corporation — Cyber Attack Analytics -----")
print("NumPy yordamida 2024 va 2025-yillardagi kiberhujumlarni tahlil qilish\n")
import numpy as np
# 2024-yildagi 12 ta hujumning tartib raqamlari
attacks_2024 = np.arange(1, 13)

# 2024-yilda har bir hujumdan kelib chiqqan shartli zarar (USD)
damage_2024 = np.array([
    1200, 4500, 800, 12000,
    3200, 7500, 2100, 18000,
    950, 6200, 14500, 3800
])

# 2025-yildagi 15 ta hujumning tartib raqamlari
attacks_2025 = np.arange(1, 16)

# 2025-yilda har bir hujumdan kelib chiqqan shartli zarar (USD)
damage_2025 = np.array([
    2500, 6800, 15000, 4200, 9500,
    22000, 1700, 8300, 12500, 3600,
    28000, 5400, 11200, 19000, 7200
])

print("2024-yil hujumlar soni:", len(attacks_2024))
print("2024-yil zararlar:", damage_2024)
print("\n")
print("2025-yil hujumlar soni:", len(attacks_2025))
print("2025-yil zararlar:", damage_2025)
print("\n##1-masala javoblari")
#----------------------------------------------------------------#

"""
1-masala. Har bir yilda nechta hujum sodir bo‘lgan?
Vazifa: Har bir yil datasetidagi hujumlar sonini aniqlang va yillar orasidagi farqni toping.
"""

count_2024 = len(attacks_2024)
count_2025 = len(attacks_2025)

print("2024-yildagi hujumlar:", count_2024)
print("2025-yildagi hujumlar:", count_2025)
print("Hujumlar sonidagi farq:", count_2025 - count_2024)
print("\n##2-masala javoblari")
#----------------------------------------------------------------#

"""
2-masala. Jami moliyaviy zararni hisoblash
Vazifa: np.sum() yordamida har bir yil uchun barcha zararlarni qo‘shing.
"""

total_2024 = np.sum(damage_2024)
total_2025 = np.sum(damage_2025)

print(f"2024-yil jami zarar: ${total_2024:,.2f}")
print(f"2025-yil jami zarar: ${total_2025:,.2f}")
print("\n##3-masala javoblari")
#----------------------------------------------------------------#

"""
3-masala. Bir hujumga to‘g‘ri keladigan o‘rtacha zarar
Vazifa: np.mean() yordamida har bir yil uchun o‘rtacha zararni hisoblang.
"""

avg_2024 = np.mean(damage_2024)
avg_2025 = np.mean(damage_2025)

print(f"2024-yil o'rtacha zarar: ${avg_2024:,.2f}")
print(f"2025-yil o'rtacha zarar: ${avg_2025:,.2f}")
print("\n##4-masala javoblari")
#----------------------------------------------------------------#

"""
4-masala. Eng katta va eng kichik zarar
Vazifa: np.max() va np.min() funksiyalari orqali har bir yilning eng katta va eng kichik zararlarini toping.
"""

print(f"2024-yil eng katta zarar: ${np.max(damage_2024):,.2f}")
print(f"2024-yil eng kichik zarar: ${np.min(damage_2024):,.2f}")

print(f"2025-yil eng katta zarar: ${np.max(damage_2025):,.2f}")
print(f"2025-yil eng kichik zarar: ${np.min(damage_2025):,.2f}")
print("\n##5-masala javoblari")
#----------------------------------------------------------------#

"""
5-masala. Qaysi yilda jami zarar ko‘proq bo‘lgan?
Vazifa: Ikki yilning jami zararini solishtirib, kattaroq qiymatga ega yilni aniqlang.
"""

if total_2025 > total_2024:
    print("2025-yilda jami zarar ko'proq bo'lgan.")
elif total_2024 > total_2025:
    print("2024-yilda jami zarar ko'proq bo'lgan.")
else:
    print("Ikkala yildagi jami zarar teng.")
print("\n##6-masala javoblari")
#----------------------------------------------------------------#

"""
6-masala. Jami zarar necha foizga o‘zgargan?
Vazifa: 2024-yilni boshlang‘ich qiymat deb olib, 2025-yildagi jami zarar foiz hisobida qancha oshgan yoki kamayganini aniqlang.
"""

change_percent = ((total_2025 - total_2024) / total_2024) * 100

print(f"Jami zarar o'zgarishi: {change_percent:.2f}%")

if change_percent > 0:
    print("Jami zarar oshgan.")
elif change_percent < 0:
    print("Jami zarar kamaygan.")
else:
    print("Jami zarar o'zgarmagan.")
print("\n##7-masala javoblari")
#----------------------------------------------------------------#

"""
7-masala. $10,000 dan yuqori zarar keltirgan hujumlar
Vazifa: Boolean indexing orqali zarari $10,000 dan yuqori bo‘lgan hujumlarni ajrating va ularning sonini hisoblang.
"""

threshold = 10_000
high_damage_2024 = damage_2024[damage_2024 > threshold]
high_damage_2025 = damage_2025[damage_2025 > threshold]

print("2024-yilda $10,000 dan yuqori zararlar:", high_damage_2024)
print("Ularning soni:", np.count_nonzero(damage_2024 > threshold))
print("2025-yilda $10,000 dan yuqori zararlar:", high_damage_2025)
print("Ularning soni:", np.count_nonzero(damage_2025 > threshold))
print("\n##8-masala javoblari")
#----------------------------------------------------------------#

"""
8-masala. Eng katta zarar keltirgan hujumning tartib raqami
Vazifa: np.argmax() yordamida eng katta zarar joylashgan indeksni toping. Indekslar 0 dan boshlangani uchun tartib raqamiga 1 qo‘shing.
"""

index_2024 = np.argmax(damage_2024)
index_2025 = np.argmax(damage_2025)

print(
    f"2024-yilda eng katta zarar {index_2024 + 1}-hujumda: "
    f"${damage_2024[index_2024]:,.2f}"
)
print(
    f"2025-yilda eng katta zarar {index_2025 + 1}-hujumda: "
    f"${damage_2025[index_2025]:,.2f}"
)
print("\n##9-Umumiy hisobot")
#----------------------------------------------------------------#

"""
9-masala. Umumiy hisobot
Vazifa: Asosiy ko‘rsatkichlarni bitta hisobotda jamlang.
"""

years = np.array([2024, 2025])
attack_counts = np.array([len(attacks_2024), len(attacks_2025)])
total_damages = np.array([np.sum(damage_2024), np.sum(damage_2025)])
average_damages = np.array([np.mean(damage_2024), np.mean(damage_2025)])
max_damages = np.array([np.max(damage_2024), np.max(damage_2025)])

print("=" * 48)
print("APTNAKE CORPORATION")
print("CYBER ATTACK ANALYTICS REPORT")
print("=" * 48)

for i in range(len(years)):
    print(f"\nYil: {years[i]}")
    print(f"Hujumlar soni: {attack_counts[i]}")
    print(f"Jami zarar: ${total_damages[i]:,.2f}")
    print(f"O'rtacha zarar: ${average_damages[i]:,.2f}")
    print(f"Eng katta zarar: ${max_damages[i]:,.2f}")

print("\nYillik jami zarar o'zgarishi:", f"{change_percent:.2f}%")
print("=" * 48)
````
