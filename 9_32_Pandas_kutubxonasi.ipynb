**24.** 'X' butun son ustuni boʻlgan DataFrame:
```python
df = pd.DataFrame({'X': [7, 2, 0, 3, 4, 2, 5, 0, 3, 4]})
```

Har bir qiymat uchun farqni oldingi nolga qaytaring.

Sizda natija quyidagicha boʻlishi kerak:

```
[1, 2, 0, 1, 2, 3, 4, 0, 1, 2]
```

Buni yangi `Y` ustuni deb nomlang.

````python
import pandas as pd
df = pd.DataFrame({'X': [7, 2, 0, 3, 4, 2, 5, 0, 3, 4]})
groups = (df['X'] == 0).cumsum()
Y = df.groupby(groups).cumcount() + 1
Y[df['X'] == 0] = 0
df = pd.DataFrame({'Y': Y})
print(df)
````

 **34**. Endi ushbu DataFrame uchun `adjacent` nomli yangi ustun yarating. Ushbu ustunda tarmoqdagi qoʻshni kvadratlarda topilgan mineʼlar soni boʻlishi kerak.

(Masalan, `(0, 0)` koordinatasi uchun yozuv boʻlgan birinchi qator uchun `(0, 1)`, `(1, 0)` va `(1) koordinatalarida qancha mine borligini hisoblang. , 1)`.)

````python
import numpy as np
import pandas as pd
X = 5
Y = 4
df = pd.DataFrame(
    pd.MultiIndex.from_product([range(X), range(Y)]).tolist(),
    columns=['x', 'y']
)
df['mine'] = np.random.binomial(1, 0.4, X * Y)
grid = df['mine'].to_numpy().reshape(X, Y)
padded = np.pad(grid, 1, constant_values=0)
adjacent = sum(
    padded[1 + dx : 1 + dx + X, 1 + dy : 1 + dy + Y]
    for dx in (-1, 0, 1)
    for dy in (-1, 0, 1)
    if (dx, dy) != (0, 0)
)
df['adjacent'] = adjacent.ravel()
print(df)
````

##📌BONUS vazifa

Oʻzingiz javob berishni istagan savolni kiriting va unga javob bering.

````python
"""
# **************************************************************************#
# ************************ Onlayn do’kon buyurtmalari **********************#
# **************************************************************************#
Quyidagi kod ikkita jadval yaratadi: orders (200 ta buyurtma, ma'lumotlar buzilgan) va customers (30 ta mijoz). Hamma mashqlar shu jadvallar ustida bajariladi.

------------------------ Vazifalar -----------------------------
1-bosqich. Tanishuv

orders hajmini, ustun turlarini, birinchi 5 qatorini va raqamli ustunlar statistikasini chiqaring.
Har bir ustunda nechta NaN borligini toping.

2-bosqich. Tozalash
3. Takroriy qatorlarni o’chiring.
4. city ustunidagi ortiqcha bo’shliq va katta-kichik harf muammosini tuzating (TOSHKENT → Toshkent).
5. date ustunini datetime turiga o’tkazing.
6. quantity dagi NaN larni medianа bilan to’ldirib, ustunni int qiling. payment dagi NaN larni 'noma'lum' bilan almashtiring.

3-bosqich. Yangi ustunlar
7. total = quantity * price ustunini qo’shing.
8. date dan month (oy raqami) va weekday (hafta kuni nomi) ustunlarini ajrating.
9. total bo’yicha size ustunini yarating: 0–100 → kichik, 100–1000 → o'rta, 1000+ → katta (pd.cut).

4-bosqich. Tanlash va saralash
10. Toshkentdagi total > 1000 bo’lgan buyurtmalarni tanlang.
11. Faqat karta yoki online bilan to’langan buyurtmalarni tanlang (isin).
12. Eng katta 5 ta buyurtmani chiqaring (nlargest).

5-bosqich. Guruhlash
13. Har bir category bo’yicha umumiy daromadni toping.
14. Har bir shahar uchun buyurtmalar soni, o’rtacha total va maksimal total ni bitta jadvalda chiqaring (agg).
15. Qator = shahar, ustun = to’lov turi, qiymat = umumiy daromad bo’lgan jadval yarating, jami qatorlari bilan (pivot_table).

6-bosqich. Jadvallarni birlashtirish
16. orders ga customers ni customer_id bo’yicha merge qiling.
17. Qaysi segment o’rtacha ko’proq xarid qilishini toping.
18. Eng ko’p pul sarflagan 5 ta mijozni (name bilan) chiqaring.

7-bosqich. Vaqt bilan ishlash
19. Oylar bo’yicha daromadni hisoblang (resample) va oldingi oyga nisbatan o’sishni foizda toping (pct_change).
20. Kunlik daromadning 7 kunlik harakatlanuvchi o’rtachasini hisoblang (rolling).

8-bosqich. Yakuniy hisobot
21. Mijozlarni yoshi bo’yicha guruhlang (18-25, 26-35, 36-60) va har bir guruh uchun buyurtmalar soni, jami daromad va eng ko’p sotilgan mahsulotni bitta jadvalda chiqaring.
"""

import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
n = 200
base_price = {'Noutbuk': 700, 'Telefon': 450, 'Monitor': 200,
              'Quloqchin': 40, 'Klaviatura': 30, 'Sichqoncha': 15}
category = {'Noutbuk': 'Elektronika', 'Telefon': 'Elektronika', 'Monitor': 'Elektronika',
            'Quloqchin': 'Aksessuar', 'Klaviatura': 'Aksessuar', 'Sichqoncha': 'Aksessuar'}
orders = pd.DataFrame({
    'order_id': np.arange(1001, 1001 + n),
    'date': (pd.to_datetime('2026-01-01')
             + pd.to_timedelta(rng.integers(0, 180, n), unit='D')).strftime('%Y-%m-%d'),
    'customer_id': rng.integers(1, 31, n),
    'product': rng.choice(list(base_price), n),
    'quantity': rng.integers(1, 6, n).astype(float),
    'city': rng.choice(['Toshkent', 'Samarqand', 'Buxoro', 'Andijon', 'Namangan'], n),
    'payment': rng.choice(['naqd', 'karta', 'online'], n),
})
orders['price'] = (orders['product'].map(base_price) * rng.uniform(0.9, 1.1, n)).round(2)
orders['category'] = orders['product'].map(category)

# Ataylab iflos qilamiz
orders.loc[rng.choice(n, 6, replace=False), 'quantity'] = np.nan
orders.loc[rng.choice(n, 8, replace=False), 'payment'] = np.nan
idx = rng.choice(n, 15, replace=False)
orders.loc[idx, 'city'] = orders.loc[idx, 'city'].str.upper() + ' '
orders = pd.concat([orders, orders.sample(5, random_state=1)], ignore_index=True)

customers = pd.DataFrame({
    'customer_id': np.arange(1, 31),
    'name': [f'Mijoz_{i}' for i in range(1, 31)],
    'age': rng.integers(18, 60, 30),
    'segment': rng.choice(['standard', 'premium'], 30),
})
print("\n")
# 1-2
print(orders.shape)
orders.info()
print(orders.describe())
print(orders.isna().sum())
print("\n")
# 3-6
orders = orders.drop_duplicates(subset='order_id').reset_index(drop=True)
orders['city'] = orders['city'].str.strip().str.title()
orders['date'] = pd.to_datetime(orders['date'])
orders['quantity'] = orders['quantity'].fillna(orders['quantity'].median()).astype(int)
orders['payment'] = orders['payment'].fillna("noma'lum")
print("\n")
# 7-9
orders['total'] = orders['quantity'] * orders['price']
orders['month'] = orders['date'].dt.month
orders['weekday'] = orders['date'].dt.day_name()
orders['size'] = pd.cut(orders['total'], bins=[0, 100, 1000, np.inf],
                        labels=['kichik', "o'rta", 'katta'])
print("\n")
# 10-12
print(orders[(orders['city'] == 'Toshkent') & (orders['total'] > 1000)])
print(orders[orders['payment'].isin(['karta', 'online'])])
print(orders.nlargest(5, 'total'))
print("\n")
# 13-15
print(orders.groupby('category')['total'].sum())
print(orders.groupby('city').agg(
    orders=('order_id', 'count'),
    avg_total=('total', 'mean'),
    max_total=('total', 'max')))
print(orders.pivot_table(index='city', columns='payment', values='total',
                         aggfunc='sum', fill_value=0, margins=True))
print("\n")
# 16-18
full = orders.merge(customers, on='customer_id', how='left')
print(full.groupby('segment')['total'].mean())
print(full.groupby('name')['total'].sum().nlargest(5))
print("\n")
# 19-20
monthly = orders.set_index('date').resample('MS')['total'].sum()
print(pd.DataFrame({'revenue': monthly, 'growth_%': monthly.pct_change() * 100}))
daily = orders.set_index('date')['total'].resample('D').sum()
print(daily.rolling(7).mean().head(10))
print("\n")
# 21
full['age_group'] = pd.cut(full['age'], bins=[17, 25, 35, 60],
                           labels=['18-25', '26-35', '36-60'])
report = full.groupby('age_group', observed=True).agg(
    orders=('order_id', 'count'),
    revenue=('total', 'sum'),
    top_product=('product', lambda s: s.mode()[0]))
print(report)
````
